# Importación de librerías

En esta sección se importan las librerías necesarias para el desarrollo del proyecto.

La librería requests permitirá realizar consultas a la API pública mediante peticiones HTTP.

La librería pandas permitirá transformar y manipular los datos obtenidos en estructuras tabulares tipo DataFrame, facilitando su limpieza y análisis posterior.

```python
import requests
import pandas as pd
```

# Definición de la API y consulta de datos mediante petición HTTP

Se define la dirección URL de la API pública que será utilizada para obtener la información.

La API seleccionada corresponde a Studio Ghibli (https://ghibliapi.vercel.app/films) y contiene datos relacionados con películas como nombre, director, duración y calificación.

Explicación:

- requests.get() hace la consulta
- json() convierte la respuesta a diccionario Python
- datos[:2] Antes de realizar transformaciones es importante inspeccionar la estructura inicial de la información obtenida. Se visualizarán los primeros registros para verificar que la consulta se realizó correctamente.

```python
url="https://ghibliapi.vercel.app/films"

respuesta=requests.get(url)

datos=respuesta.json()

datos[:2]
```

# Conversión de JSON a DataFrame

La información proveniente de la API se encuentra en formato JSON.

Para facilitar el análisis y procesamiento de los datos, estos serán convertidos a un DataFrame utilizando la librería pandas.



```python
df=pd.DataFrame(datos)
```

# Visualización inicial del DataFrame

Se muestran las primeras filas del conjunto de datos para identificar la estructura y contenido de las variables disponibles.


```python
df.head()
```

# Exploración de variables disponibles

En esta etapa se identifican las columnas existentes dentro del conjunto de datos para seleccionar únicamente aquellas que serán útiles para el análisis y visualización.

```python
df.columns
```

# Selección de variables relevantes

Se seleccionan únicamente las variables que aportarán valor al análisis:

- Título
- Director
- Productor
- Año
- Duración
- Calificación

Esto permite trabajar con un conjunto de datos más limpio y enfocado.

```python
df=df[
[
'title',
'director',
'producer',
'release_date',
'running_time',
'rt_score'
]
]
```

# Visualización de la nueva estructura

Después de filtrar las variables necesarias se revisan nuevamente los datos para confirmar los cambios realizados.

```python
df.head()
```

# Transformación de tipos de datos

Algunas variables numéricas se encuentran almacenadas como texto.

Para poder realizar cálculos y visualizaciones correctamente es necesario convertirlas a formato numérico.

```python
df['release_date']=pd.to_numeric(df['release_date'])

df['running_time']=pd.to_numeric(df['running_time'])

df['rt_score']=pd.to_numeric(df['rt_score'])
```

# Renombramiento de columnas

Se modifican los nombres originales de las variables por nombres más descriptivos y fáciles de interpretar dentro de Power BI.

```python
df.columns=[
'Titulo',
'Director',
'Productor',
'Año',
'Duracion_Min',
'Calificacion'
]
df.head()
```

# Creación de una nueva variable categórica

Se creará una nueva variable denominada Tipo_Duración.

Esta variable clasificará las películas según su duración:

- Corta
- Media
- Larga

Esto facilitará posteriores análisis y filtros en Power BI.

```python
def categoria_duracion(x):

    if x<90:
        return "Corta"

    elif x<120:
        return "Media"

    else:
        return "Larga"

df["Tipo_Duracion"]=df["Duracion_Min"].apply(categoria_duracion)
```

# Visualización del conjunto de datos transformado

Se revisan nuevamente los datos para comprobar la correcta creación de la nueva variable y validar la estructura final.

```python
df.head()
```

# Verificación de valores faltantes

Es importante verificar si existen datos nulos o vacíos dentro del conjunto de datos.

La presencia de valores faltantes podría afectar los análisis y visualizaciones posteriores.

```python
df.isnull().sum()
```

# Exportación del conjunto de datos

Finalmente se exportará el conjunto de datos procesado y limpio a un archivo CSV.

Este archivo será utilizado posteriormente en Power BI para la construcción del dashboard interactivo.

```python
df.to_csv(
"peliculas_ghibli.csv",
index=False
)

print("Archivo CSV creado correctamente")
```
